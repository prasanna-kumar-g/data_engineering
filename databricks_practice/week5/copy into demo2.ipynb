```python
%sql
drop table if exists orders;
create table orders(order_id int,
order_date string,
customer_id int,
order_status string,
file_name string,
loaded_date timestamp)
using delta;
```

```python
%sql
copy into orders 
from (select cast(order_id as int) as order_id,order_date ,cast(customer_id as int)as customer_id,order_status,_metadata.file_name as file_name,current_timestamp() as loaded_date 
from '/Volumes/workspace/default/orders_datasets/orders/')
fileformat=csv
format_options(
  "header"="true"
)
```
