### enhancing the functionality of function without changing the function 

```python
def greet():
  print("hello")
greet()
```

```python
def my_decorator(func):
    def wrapper():
        print('before calling function')
        func()
        print("after calling function")
    return wrapper
my_decorator(greet)()

```

```python
@my_decorator
def greet():
  print("hello")
greet()
```

### read the data frm cloud and write it to delta table

```python
@write_to_delta
def orders_bronze_latest():
  df=(
    spark.read
    .format("csv")
    .option("header","true")
    .option("inferSchema","true")
    .load("/Volumes/dlt/dlt_practice/dlt_data/orders")
  )

  return df
```

```python
def write_to_delta(func):
    def wrapper():
        df=func()
        table_name=func.__name__
        df.write.format("delta").mode("append").saveAsTable("dlt.dlt_practice."+table_name)
    return wrapper
```
